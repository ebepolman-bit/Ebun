````python
# Hier de data en de analyse
import numpy as np
import matplotlib.pyplot as plt

Voltage = 12.5 #in volt
Ampere = 1.60 #in ampere
T_s = 23.2 #in c
T = np.linspace(T_s, T_s + 5, 51)
gemeten_dt = np.array([0,9.61,6.32,7.53,7.32,9.25,7.53,7.43,7.37,9.4,7.58,7.23,7.38,9.28,7.5,7.43,8.31,8.38,8.31,8.45,7.4,9.16,7.52,8.28,8.48,8.28,8.33,7.4,2.87,1.99,23.54,7.13,8.2,9.18,8.4,8.31,8.45,9.16,8.56,8.25,8.45,9.23,8.13,8.5,7.63,10.15,9.16,8.21,8.93,9.91,8.18])

# punten die ver afwijken van de rest eruit halen
gemeten_dt_filtered = gemeten_dt[((gemeten_dt > 5)*(gemeten_dt< 15))]

plt.figure
plt.plot(T[((gemeten_dt > 5)*(gemeten_dt< 15))],gemeten_dt_filtered, "b.")
plt.savefig("dt_per_T.png", dpi=450)

dt = np.mean(gemeten_dt_filtered)
print("average measured time per 0.1 degree celsius",dt)
dt10 = dt * 10
Wattage = Voltage*Ampere
cv = (dt*10*Wattage)/(0.4)
print("De gevonden waarde voor c_water is", round(cv, 1), "J/K*kg")
````
